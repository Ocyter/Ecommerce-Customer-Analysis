# E-commerce Customer Analysis — Signal Map

A compact, portfolio-oriented walkthrough of customer behavior across device, location and purchase activity.

```python
import pandas as pd
df = pd.read_csv('ecommerce_customer_data.csv')
df.head()
```

## 01 — What does the customer base look like?

We have 500 customers, ages 18–35, across 8 locations and 3 device types.

```python
df.describe(include='all').T
```

## 02 — Where is purchase activity strongest?

Compare average purchases and purchase rate by location rather than treating every customer as one homogeneous audience.

```python
location = df.groupby('Location').agg(
    Customers=('User_ID','count'),
    Avg_Purchases=('Total_Purchases','mean'),
    Purchase_Rate=('Total_Purchases', lambda x: (x > 0).mean())
).sort_values('Avg_Purchases', ascending=False)
location.round(3)
```

## 03 — The device signal

Mobile is the largest device segment and also records the highest average purchases, although the gap is small.

```python
df.groupby('Device_Type')['Total_Purchases'].agg(['count','mean']).sort_values('mean', ascending=False).round(2)
```

## 04 — The surprising result

Browsing time, pages viewed and cart items have almost zero Pearson correlation with purchase volume in this sample. That is a useful finding, not a failure: it suggests simple engagement volume is not enough to explain purchase behavior here.

```python
df[['Age','Product_Browsing_Time','Total_Pages_Viewed','Items_Added_to_Cart','Total_Purchases']].corr()['Total_Purchases'].round(3)
```
